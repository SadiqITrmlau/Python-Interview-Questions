## Find Largest Number in the list using For Loop.

```python
num_list=[1,23,12,3,4,12,4,24,12,12,12,36]
```

```python
# linear seacrh or brute-force approach

#by default assume the no. on 1st index is largest
largest=num_list[0]
for num in num_list:
    if num>largest:
        largest=num
print("Largest no. is: ",largest)
```

### Edge Cases:-
1. What if all numbers are equal?
2. What if the list has only one element?
3. What if the list is empty?
4. What if Input is not a list?
5. A Non-Numeric value.

```python
def find_largest(num_list: list[float])->float:
    """
    Function to find largest number in the list.
    Args: 
        num_list(num_list: list[float]): A list containing numeric values (int or float).

    Returns: 
        float: The largest number in the list

    Raises: 
        TypeError: if input in not a list or contains a non-numeric values.
        ValueError: If the list is Empty
    """

    # Validate that input is a list
    if not isinstance(num_list,list): # agr num_list object nahi hai list class ka 
        raise TypeError("Input must be a list of int/float.")
    # Validate that list is not empty 
    if not num_list:
        raise ValueError("List cannot be empty.")
    # Validate that list elements should be numeric 
    for num in num_list:
        if not isinstance(num,(int,float)):
            raise TypeError("All elements must be numeric.")
    largest=num_list[0]
    for num in num_list:
        if num>largest:
            largest=num
    return largest

# ---Main Execution Part---
if __name__=="__main__":
    try:
        # taking input from user
        user_input=input("Enter numbers in the list separated by space of int/float only: ")
        # converting user_input into list
        num_list=[float(item) for item in user_input.split()]
        # calling the function
        result=find_largest(num_list)
        # display the result
        print("Largest number in the list is: ",result)
    except (TypeError,ValueError) as error:
        print(error)
    except Exception as e:
        print(e)
```

```python
find_largest([1,2,"4b",8])
```
